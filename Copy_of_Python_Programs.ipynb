1.Write a Python program to demonstrate how to Draw a Bar Plot using Mat plot lib.

```python
import matplotlib.pyplot as plt

#Giving the data to plot
courses = ['Python','Java','C++','Javascript','AI']
students = [50,40,30,60,45]

#Ploting bar graph
plt.bar(courses,students)

#Specifying the title and labels
plt.title('Numbers of students enrolled in each course')
plt.xlabel('courses')
plt.ylabel('Number of students')

#Shows the bar graph
plt.show()
```

2.Write aPython program to demonstrate how to Draw a Scatter Plot using Mat plot lib

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50]
y = [5,15,25,10,30]

plt.scatter(x,y)
plt.title("Sample scatter plot")
plt.xlabel("x values")
plt.ylabel("y values")
plt.show()
```

3.Write a Python program to demonstrate how to Draw a Histogram Plot using Mat plot lib.

```python
import matplotlib.pyplot as plt

ages = (12,23,45,22,36,27,29,40,50,30,28,33,31,18,21,25)

plt.hist(ages,bins=5)
plt.title("Age Distribution Histogram")
plt.xlabel("Age Range")
plt.ylabel("Frequency")
plt.show()
```

4.Write a Python program to demonstrate how to Draw a Pie Chart using Mat plot lib.

```python
import matplotlib.pyplot as plt

courses = ['Python','Java','C++','AI']
students = [40,25,20,15]

plt.pie(students, labels=courses, autopct='%1.1f%%')
plt.title("Students Enrollment in Different Courses")
plt.show()
```

6.Write a Python program to illustrate Stack and subplot usingMatplotlib.

```python
import matplotlib.pyplot as plt

days = [1,2,3,4,5]

sleeping = [7,8,6,7,8]
working = [8,9,7,8,9]
playing = [2,3,4,3,2]

fig, (ax1, ax2) = plt.subplots(1,2, figsize=(10,4))

ax1.stackplot(days,sleeping,working,playing, labels=['Sleeping', 'Working', 'Playing'])
ax1.set_title("Daily Activity Stack Plot")
ax1.set_xlabel("Days")
ax1.set_ylabel("Hours")
ax1.legend(loc='upper left')

ax2.plot(days, sleeping, label='Sleeping')
ax2.plot(days, working, label='Working')
ax2.plot(days, playing, label='Playing')
ax2.set_title("Subplot: Activity Lines")
ax2.set_xlabel("Days")
ax2.set_ylabel("Hours")
ax2.legend()

plt.tight_layout()
plt.show()
```
